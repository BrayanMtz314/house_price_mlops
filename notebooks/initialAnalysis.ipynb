# Looking for outliers and unusual values

```python
def plot_boxplot(column_name):
    fig, ax = plt.subplots(figsize=(10, 6))
    bp = ax.boxplot(df[column_name])
    ax.set_title(f'Boxplot of {column_name}')
    ax.set_ylabel(column_name)
    plt.show()
    
for column in df.columns:
    plot_boxplot(column)
```

```python
def plot_histogram(column_name):
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.hist(df[column_name], bins=30, edgecolor='black')
    ax.set_title(f'Histogram of {column_name}')
    ax.set_xlabel(column_name)
    ax.set_ylabel('Frequency')
    plt.show()

for column in df.columns:
    plot_histogram(column)
```
