```python
import sys
from pathlib import Path

current = Path.cwd()

repo_root = next(
    (
        path
        for path in [current, *current.parents]
        if (
            path / "src" / "house_price_simple_analysis"
        ).is_dir()
    ),
    None,
)

if repo_root is None:
    raise FileNotFoundError(
        f"Project root not found from {current}"
    )

src_path = str(repo_root / "src")

if src_path not in sys.path:
    sys.path.insert(0, src_path)

from house_price_simple_analysis.config import ProjectConfig

config = ProjectConfig.from_yaml(
    str(repo_root / "config" / "project_config.yml")
)
```

```python
from house_price_simple_analysis.model_lifecycle import ModelLifecycle

# Retrieve values from previous Job tasks
candidate_uri = dbutils.jobs.taskValues.get(
    taskKey="train_model",
    key="candidate_model_uri"
)

approved = dbutils.jobs.taskValues.get(
    taskKey="validate_model",
    key="candidate_approved"
)

print("Candidate URI:", candidate_uri)
print("Approved:", approved)
```

```python
if approved is True:
    lifecycle = ModelLifecycle(config=config)

    registered_uri = lifecycle.register_candidate(
        model_uri=candidate_uri
    )

    print("Candidate registered successfully.")
    print("Registered URI:", registered_uri)

    dbutils.jobs.taskValues.set(
        key="registered_candidate_uri",
        value=registered_uri
    )

else:
    print("Candidate rejected. Registration skipped.")
```
