# Testing model deployment class

This class was done to manage deployed models. Inspects the registered model, checks the serving endpoint, and sends inference requests.

### Inicialize object and get the champion version

```python
import sys
from pathlib import Path

# Find the repository root from the notebook's directory
current = Path.cwd()

repo_root = next(
    (
        path
        for path in [current, *current.parents]
        if (path / "src" / "house_price_simple_analysis").is_dir()
    ),
    None
)

if repo_root is None:
    raise FileNotFoundError(
        "Could not locate src/house_price_simple_analysis. "
        f"Current directory: {current}"
    )

sys.path.insert(0, str(repo_root / "src"))

print("Repository root:", repo_root)
print("Python source directory:", repo_root / "src")
```

```python
from house_price_simple_analysis.model_deployment import ModelDeployment
from house_price_simple_analysis.config import ProjectConfig

config = ProjectConfig.from_yaml(
    str(repo_root / "config" / "project_config.yml")
)

deployment = ModelDeployment(
    config=config,
    endpoint_name="house-price-endpoint"
)
```

```python
# Retrieve the Champion version
version = deployment.get_champion_version()

print("Champion version:", version)
```

### Inspect the endpoint

```python
endpoint = deployment.get_endpoint()

print("Endpoint:", endpoint.name)
print("State:", endpoint.state)
print("Served entities:", endpoint.config.served_entities)
```

### Sending a prediction

```python
prediction = deployment.predict(2000.0)

print("Predicted house price:", prediction)
```
